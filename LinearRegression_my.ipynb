# Linear Regression

Let's fabricate some data that shows a roughly linear relationship between page speed and amount purchased:

```python
import matplotlib.pyplot as plt
from sympy.stats.sampling.sample_scipy import scipy
%matplotlib inline
import numpy as np
from pylab import *

page_speeds = np.random.normal(3.0, 1.0, 1000)
# np.histogram(page_speeds)
plt.hist(page_speeds, bins=50, color='r', edgecolor='k')
plt.xlabel('Page speed')
plt.ylabel('Frequency')
plt.title('Histogram of Page Speeds')
plt.show()
```

```python
purchase_amount = 100 - (page_speeds + np.random.normal(0.0, 0.09, 1000))*3

plt.hist(purchase_amount, bins=50, color='b', edgecolor='k')
plt.xlabel('Purchase amount')
plt.ylabel('Frequency')
plt.title('Histogram of Purchase Amount')
plt.show()
```

```python

scatter(page_speeds, purchase_amount)
```

As we only have two features, we can keep it simple and just use scipy.state.linregress:

```python
# import scipy.stats
from scipy import stats

# slope, intercept, r_value, p_value, std_err = stats.linregress(page_speeds, purchase_amount)

result = stats.linregress(page_speeds, purchase_amount)
slope, intercept, r_value, p_value, std_err, intercept_stderr = (
    result.slope, result.intercept, result.rvalue, result.pvalue, result.stderr, result.intercept_stderr
)

print(f"slope:{result.slope}, intercept:{result.intercept}, rvalue:{result.rvalue}, pvalue:{result.pvalue}, stderr:{result.stderr}, intercept_stderr:{result.intercept_stderr}")
```

Not surprisngly, our R-squared value shows a really good fit:

```python
r_squared = result.rvalue ** 2
print(r_squared)
```

Let's use the slope and intercept we got from the regression to plot predicted values vs. observed:

```python

def prediction_line(x):
    return result.slope * x + intercept

fitline = prediction_line(page_speeds)

#draw a scatter plot for page_speed and purchase_amount
plt.scatter(page_speeds, purchase_amount)
plt.plot(page_speeds, fitline, c='r')
plt.show()
```
