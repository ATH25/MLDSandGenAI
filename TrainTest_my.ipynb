# Train / Test

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Constants
SEED = 2
PAGE_SPEED_MEAN = 3.0
PAGE_SPEED_STDDEV = 1.0
PURCHASE_AMOUNT_MEAN = 50.0
PURCHASE_AMOUNT_STDDEV = 30.0
DATA_SIZE = 100


def generate_data(seed=SEED, size=DATA_SIZE,
                  page_speed_mean=PAGE_SPEED_MEAN, page_speed_std=PAGE_SPEED_STDDEV,
                  purchase_mean=PURCHASE_AMOUNT_MEAN, purchase_std=PURCHASE_AMOUNT_STDDEV):
    """Generates synthetic page speed and purchase amount data."""
    np.random.seed(seed)
    page_speeds = np.random.normal(page_speed_mean, page_speed_std, size)
    purchase_amount = np.random.normal(purchase_mean, purchase_std, size) / page_speeds
    return page_speeds, purchase_amount

def plot_data(x, y, xlabel='Page Speeds', ylabel='Purchase Amount', title='Scatter Plot'):
    """Displays a scatter plot of x vs y."""
    plt.scatter(x, y)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()

# Generate and plot
pageSpeeds, purchaseAmount = generate_data()
plot_data(pageSpeeds, purchaseAmount)
```

Now we'll split the data in two - 80% of it will be used for "training" our model, and the other 20% for testing it. This way we can avoid overfitting.



```python
trainX = pageSpeeds[:80]
testX = pageSpeeds[80:]

trainY = purchaseAmount[:80]
testY = purchaseAmount[80:]
```

Here's our training dataset:

```python

plot_data(trainX, trainY)
```

And our test dataset:

```python
# plt.scatter(testX, testY)
plot_data(testX, testY)
```

Now we'll try to fit an 8th-degree polynomial to this data (which is almost certainly overfitting, given what we know about how it was generated!)

```python
x = np.array(trainX)
y = np.array(trainY)

p4 = np.poly1d(np.polyfit(x, y, 8))
# print(np.polyfit(x, y, 8))
# print(p4)
```

Let's plot our polynomial against the training data:

```python
import matplotlib.pyplot as plt

xp = np.linspace(0, 7, 100)
axes = plt.axes()
axes.set_xlim([0,7])
axes.set_ylim([0, 200])
plt.scatter(x, y)
plt.plot(xp, p4(xp), c='r')
plt.show()
```

And against our test data:

```python
testx = np.array(testX)
testy = np.array(testY)

axes = plt.axes()
axes.set_xlim([0,7])
axes.set_ylim([0, 200])
plt.scatter(testx, testy)
plt.plot(xp, p4(xp), c='r')
plt.show()
```

Doesn't look that bad when you just eyeball it, but the r-squared score on the test data is kind of horrible! This tells us that our model isn't all that great...

```python
from sklearn.metrics import r2_score

r2 = r2_score(testy, p4(testx))

print(r2)
```

...even though it fits the training data better:

```python
from sklearn.metrics import r2_score

r2 = r2_score(np.array(trainY), p4(np.array(trainX)))

print(r2)
```

If you're working with a Pandas DataFrame (using tabular, labeled data,) scikit-learn has built-in train_test_split functions to make this easy to do.

Later we'll talk about even more robust forms of train/test, like K-fold cross-validation - where we try out multiple different splits of the data, to make sure we didn't just get lucky with where we split it.
