Now let's make purchaseAmount inversely proportional to pageSpeeds

purchaseAmount = np.random.normal(50.0, 10.0, 1000) / pageSpeeds

Logic Behind the Division

The logic behind dividing by pageSpeeds is to create an inverse relationship between purchaseAmount and pageSpeeds.
	1.	What This Means:
	•	If pageSpeeds (page load time) is high, the result of purchaseAmount / pageSpeeds will be low.
	•	If pageSpeeds is low, the result of purchaseAmount / pageSpeeds will be high.
	•	This enforces the idea that faster pages lead to higher purchases.
	2.	Why Create an Inverse Relationship?
	•	The dataset is being artificially structured to demonstrate negative correlation (higher page speed → lower purchase amount).
	•	This is useful for illustration purposes, so when computing covariance and correlation, we can see a clear negative relationship.

Example Calculation

Assume we generate:

pageSpeeds = np.array([2.0, 3.0, 4.0])
purchaseAmount = np.random.normal(50.0, 10.0, 3)  # Example: [55.0, 45.0, 60.0]

Now, after division:

purchaseAmount = [55.0 / 2.0, 45.0 / 3.0, 60.0 / 4.0]
              = [27.5, 15.0, 15.0]

So, as pageSpeeds increases, purchaseAmount decreases.

Takeaway
	•	This division inverts the relationship.
	•	It ensures that pages that load slowly correspond to lower purchases, and fast pages correspond to higher purchases.
	•	This simulates a negative correlation, which we later validate using covariance and correlation calculations.



```python

purchaseAmount = np.random.normal(50.0, 10.0, 1000) / pageSpeeds

plt.scatter(pageSpeeds, purchaseAmount, edgecolor='k', color='r', label="Purchase Amount")
plt.title("Scatter Plot of Page Speeds vs Purchase Amount")
plt.xlabel("Page Load Speed (seconds)")
plt.ylabel("Purchase Amount")
plt.legend()  # Show the legend
plt.show()
```

```python
# Write a covariance function and call covariance using pageSpeeds and purchaseAmount
# covariance = np.cov(pageSpeeds, purchaseAmount)
# print(covariance)
# print(covariance[0, 1])

def de_mean(x):
    x_bar = np.mean(x)
    return [xi-x_bar for xi in x]

def covariance(x, y):
    dot_product = np.dot(de_mean(x), de_mean(y))/(len(x)-1)
    return dot_product

covariance(pageSpeeds, purchaseAmount)
```

```python
#correlation
def correlation(x, y):
    # std_x = np.std(x)
    # std_y = np.std(y)
    std_x = x.std(ddof=1)
    std_y = y.std(ddof=1)
    # std_x = np.std(x, ddof=1)
    # std_y = np.std(y, ddof=1)
    return covariance(x, y) / (std_x * std_y)

correlation(pageSpeeds, purchaseAmount)

```

We can force a perfect correlation by fabricating a totally linear relationship (again, it's not exactly -1 just due to precision errors, but it's close enough to tell us there's a really good correlation here).

	•	Here, instead of using a randomly generated purchaseAmount (as in previous cases), we explicitly define purchaseAmount as:

     purchaseAmount = 100 - (pageSpeeds * 3)

	•	This creates a perfect negative correlation because as pageSpeeds increases, purchaseAmount decreases in a completely predictable way.

```python
purchaseAmount = 100 - pageSpeeds * 3
plt.scatter(pageSpeeds, purchaseAmount, edgecolor='k', color='r', label="Purchase Amount")
correlation (pageSpeeds, purchaseAmount)
```
